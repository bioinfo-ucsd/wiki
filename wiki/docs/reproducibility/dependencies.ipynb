# Dependencies

## Software (almost always) requires other software to work properly

Software packages almost always depend on other other software packages to work properly. The *dependencies* of a package refers to the software that the package requires.

For example, to install python, your computer will need a C compiler to be installed first. In turn, the C compiler will rely on certain software available through your computer's operating system.

The dependencies of a package often must be installed before you can install the package itself.

## Certain versions of dependencies will be incompatible

Not all versions of every dependency are compatible with each other. In fact, many have strict compatibility requirements. For example, numpy 1.26.0 is only compatible with `python >= 3.9`.

So what would happen if you tried to install another software package that is only compatible with python 3.7? You would encounter a *dependency conflict*.

In practice, nobody performs this task manually. Instead, data scientists use a tool called a *package manager* to automatically resolve dependency versions for them and install all of their software.

However, many data scientists don't even know how their package manager works! This can lead to frustration when it becomes impossible to find a set of compatible versions for their dependencies. It also hampers their ability to make their work reproducible.
